## Streaming and Batches

### Streaming

Most models can stream their output content while it is beign generated. By displaying ouput progressively, streaming significantly improves user exerience, particularly for longer responses. Calling `stream()` returns an iterator that yields output chunks as they are produced. You can use aloop to process each chunk in real-time.

```python
for chunk in model.stream("Write me a 200 word paragraph on Lord of the Rings"):
		print(chunk.text, end="", flush=True)
```

### Batch

Batching a collection of indeoendent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel

```python
responses = model.batch([
	"What is the capital of France?",
	"What is the capital of Germany?",
	"What is the capital of Italy?"
])

for response in responses:
	print(response.content)
```

```python
responses = model.batch([
	"What is the capital of France?",
	"What is the capital of Germany?",
	"What is the capital of Italy?",
],
	config = {
		'max_concurrency': 5
	}
)

for response in responses:
	print(response.content)
```
